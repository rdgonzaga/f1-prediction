Predicted finishing order with win, podium and points chances. Run after qualifying.
`REFRESH = True` downloads the weekend's sessions first; leave it off while a bulk `fetch` is running.

```python
SEASON = 2026
EVENT = "14"
REFRESH = False
```

```python
import matplotlib.pyplot as plt

from f1pred import plotting, predict

plotting.apply_theme()
pred = predict.run(SEASON, EVENT, refresh=REFRESH)
```

```python
top = pred.head(10).iloc[::-1]
fig, axes = plt.subplots(1, 2, figsize=(10, 4.8), sharey=True)
for ax, (col, title) in zip(axes, [("WinPct", "Win chance (%)"), ("PodiumPct", "Podium chance (%)")]):
    ax.barh(top["Abbreviation"], top[col], height=0.6, color=plotting.ACCENT)
    for y, value in enumerate(top[col]):
        ax.annotate(f"{value:.0f}%", (value, y), xytext=(4, 0), textcoords="offset points",
                    va="center", fontsize=9, color=plotting.INK_SECONDARY)
    ax.set(title=title, xlim=(0, 100))
    ax.grid(axis="y", visible=False)
plt.show()
```
